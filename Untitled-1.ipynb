```python
arr = np.array([10, 20, 30, 40, 50])
values = np.array([15, 25, 35, 45])
newsear = np.searchsorted(arr , values)
left = np.searchsorted(arr, values, side="left")
right = np.searchsorted(arr, values, side="right")
print(newsear)
print(left)
print(right)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
values = np.array([15, 25, 35, 45])

positions = np.searchsorted(arr, values)

print("Positions:", positions)

new_arr = np.insert(arr, positions, values)

print("Final array:", new_arr)
```
