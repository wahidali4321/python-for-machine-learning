## Assignment 1 — Weight & Volume → CO₂

```python
import pandas as pd
from sklearn import linear_model

data = {
    'Weight': [790, 1160, 929, 865, 1140, 929, 1109, 1365],
    'Volume': [1000, 1500, 1500, 1400, 1600, 1600, 2000, 2000],
    'CO2': [99, 110, 115, 105, 120, 115, 130, 140]
}

df = pd.DataFrame(data)
x = df[['Weight' , 'Volume']]
y = df['CO2']
regr = linear_model.LinearRegression()
regr.fit(x,y)
pred = regr.predict([[1200,1800]])
print(pred)
```
